# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

Objective:

This research aims to analyze the impact of the Russia-Ukraine war on flight behavior between Europe and Asia. It will focus on changes in flight routes and frequencies, with a particular emphasis on adjustments made by airlines and the broader consequences on air travel between these regions. The study will explore both immediate and longer-term disruptions caused by the war. 

Research Questions:

The primary research questions for this study is: 
- How has the Russia-Ukraine war influenced Euro-Asian flight behaviour, specifically regarding rerouting of flights, flight frequencies and flight specifications? 

Subquestions: 

- Have airlines diverted flights to avoid conflict zones (such as Russian airspace)? 
- What alternative routes are being used, and how do they affect travel time, distances and fuel consumption? 
- Have airlines reduced the frequency of flights due to safety concerns or declining demand? 
- How has the war influenced the number of airlines operating on these routes?

# Data Used

- OpenSky Network: Flight tracking data for historical and real-time flight routes and movements between Europe and Asia. 

- FlightRadar24: Detailed flight tracking data with a focus on routes, flight times traffic analysis and GPS jamming. 

- CarbonSpace: Data on emissions regarding flights and air travel  

- OAG Aviation: Data on flight frequencies and scheduling between European and Asian airports. 

- News Sources and Reports: Government regulations, sanctions, and political statements influencing air travel. 

# Data Pipeline

The data analysis pipeline will follow these steps: 

1) Data Collection: 

    a. Flight Route Data: Collect flight path and route information from flight tracking platforms. 

    b. Flight Frequency Data: Gather data on flight frequencies from airline schedules or industry reports. 

    c. Geopolitical Data: Track government regulations, airspace closures, and sanctions affecting flights between the regions. 

2) Data Preprocessing: 

    a. Clean and normalize datasets to ensure consistency across sources. 

    b. Filter flight data to focus on European-Asian routes and relevant time frames (from the start of the war to the present). 

3) Analysis: 

    a. Route Analysis: Use spatial data analysis to visualize and compare flight routes before and after the war. Identify changes in airspace usage, new flight routes, and route detours. 

    b. Trend Analysis: Use time series analysis to track changes in flight frequencies, identifying significant shifts or disruptions. 

    c. Correlation Analysis: Analyze the relationship between geopolitical events and changes in flight behavior. 

4) Visualization: 

    a. Use geographic information systems (GIS) and data visualization tools to illustrate flight route changes and operational adjustments. 

    b. Create interactive maps and time series plots to convey the impact of the war on air travel between Europe and Asia.
